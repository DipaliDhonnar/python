```python
# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python
#Get the shape of the matrix.
data.shape
```

```python
#average age
np.mean(data[:,0])
```

```python
#extract all math marks
data[:,1]
```

```python
#Find the highest Science mark.
np.max(data[:,2])
```

```python
#Get details of the student who scored more than 90 in Math.
data_bool=data[:,1]>90
print(data_bool)
data[data_bool]
```

```python
#Increase Math marks of all students by 5.
data[:,1]+5
```

```python
#Find how many students are younger than 19.
data_bool=data[:,0]>=19
print(data_bool)
len(data[data_bool])
```

```python
#Calculate the average marks in each subject (column-wise mean).
print(data)
math_avg=np.mean(data[:,1])
print(f"math average is {math_avg}")
scie_avg=np.mean(data[:,2])
print(f"science average is {scie_avg}")
```

```python
np.mean(data[:,1:],axis=0)
```

```python
#Get data of students who scored at least 80 in both subjects.
math_over_80 = data[:,1] >= 80
science_over_80 = data[:,2] >= 80
students_both_over_80 = data[math_over_80 & science_over_80]
print(students_both_over_80)
```
