```python
#Add a new column 'Discount' where each passenger gets a 10% discount on their ticket price.
df["Discount"] = (df['TicketPrice']*10)/100
df.head()
```

```python
#Create a new column 'FinalPrice' which is TicketPrice - Discount.
df['FinalPrice']=df["TicketPrice"]-df["Discount"]
df.head(2)
```

```python
#Create a new column 'AgeGroup' with categories:o	'Young' (Below 30)o	'Adult' (30-45)o	'Senior' (Above 45).
df['AgeGroup'] = df['Age'].apply(lambda x: 'Senior' if x > 45 else ('Adult' if 30 <= x <= 45 else 'Young'))
df.head(2)
```

```python
#Create a new column 'HighValue' with 'Yes' if TicketPrice > ₹1000, else 'No'.
df["HighValue"]=df["TicketPrice"].apply(lambda i :("Yes") if i>1000 else("No"))
```

```python
#	Find the average ticket price per seat class.
df.groupby("SeatClass")["TicketPrice"].mean()
```

```python
#Count the number of passengers in each 'AgeGroup'.
df["AgeGroup"].value_counts()
```

```python
# Find the highest ticket price in each seat class.
df.groupby("SeatClass")["TicketPrice"].max()
```

```python
#Find the lowest ticket price in each seat class.
df.groupby("SeatClass")["TicketPrice"].min()
```

```python
#Find the passenger with the highest age.
df.nlargest(1,"Age")
```

```python
#Count the number of passengers with a 'Confirmed' booking.
df[df["Status"]=="Confirmed"]["PassengerID"].count()
```

```python
#Find the average ticket price of passengers in Business Class.
df[df["SeatClass"]=="Business"]["TicketPrice"].mean()
```

```python
#Find the highest ticket price for each flight.
df.groupby('FlightNumber')['TicketPrice'].max()
```

```python
#Remove the 'Gender' column from the dataset.
df.drop("Gender" ,axis=1)
```

```python
#Reset the index of the dataset.
df = df.reset_index()
df
```

```python
#Replace all 'On Hold' statuses with 'Pending'.
df["Status"]=df["Status"].replace("On Hold","Pending")
df.tail(5)
```

```python
	#Replace all ticket prices below ₹600 with 'Low Price'.
 #df["TicketPrice"]=df["TicketPrice"].apply(lambda i :"Low Price" if i<600  else i)
```

```python
#Convert the 'TicketPrice' column to an integer type.
df["TicketPrice"]=df["TicketPrice"].astype(int)
df.dtypes
```

```python
#Create a new column 'AboveAverage' with 'Yes' if the ticket price is above the dataset's average.
df['AboveAverage']=df["TicketPrice"].apply(lambda i : "Yes" if i>df["TicketPrice"].mean() else "No")
df.head()
```

```python
#Convert the 'SeatClass' column to a categorical data type.
df["SeatClass"]=df["SeatClass"].astype("category")
df.dtypes
```

```python
#Find the top 5 passengers with the highest ticket price.
df.nlargest(5,"TicketPrice")
```

```python
#Find the bottom 5 passengers with the lowest ticket price.
df.nsmallest(5,"TicketPrice")
```
