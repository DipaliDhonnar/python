```python
def extract_episode(txt):
    check = False
    data = ""
    for i in txt:
        if i == ")":
            check = False
            return data
        if check == True:
            data = data + i
        if i == '(':
            check = True

```

```python
df["episodes"]=df["Title"].apply(extract_episode)
```

```python
#make a new column for time stamp
def extract_time_stamp(txt):
  check =False
  time=""
  for i in txt:
    if i ==")":
      check=True
    if check==True:
      time=time+i
  return(time[1:20])
```

```python
df["time_stamp"]=df["Title"].apply(extract_time_stamp)
df.head()
```

```python
month_to_num = {
    'Jan': 1, 'Feb': 2, 'Mar': 3, 'Apr': 4, 'May': 5, 'Jun': 6,
    'Jul': 7, 'Aug': 8, 'Sep': 9, 'Oct': 10, 'Nov': 11, 'Dec': 12
}

def calculate_months_between(timestamp_str):
    parts = timestamp_str.split(' - ')
    if len(parts) != 2:
        return None # Handle malformed timestamps

    start_part = parts[0].strip()
    end_part = parts[1].strip()

    start_month_str, start_year_str = start_part.split(' ')
    end_month_str, end_year_str = end_part.split(' ')

    start_month_num = month_to_num.get(start_month_str, None)
    end_month_num = month_to_num.get(end_month_str, None)
    start_year = int(start_year_str)
    end_year = int(end_year_str)

    if start_month_num is None or end_month_num is None:
        return None # Handle unknown month abbreviations

    # Calculate total months inclusive of start and end month
    total_months = (end_year - start_year) * 12 + (end_month_num - start_month_num) + 1
    return total_months

df['num_months'] = df['time_stamp'].apply(calculate_months_between)
display(df.head())
```
