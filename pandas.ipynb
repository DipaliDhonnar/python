```python
#mergiing in Pandas

employees = pd.DataFrame({
    "Emp_ID": [101, 102, 103, 104, 105],
    "Name": ["Amit", "Sneha", "Rahul", "Priya", "Akash"],
    "Dept_ID": [1, 2, 1, 3, 4]
})

print(employees)
```

```python
departments = pd.DataFrame({
    "Dept_ID": [1, 2, 3, 5],
    "Department": ["IT", "HR", "Finance", "Marketing"],
    "Location": ["Pune", "Mumbai", "Nashik", "Delhi"]
})

print(departments)
```

```python
pd.merge(employees,departments,on="Dept_ID")
```

```python
pd.merge(employees,departments,on="Dept_ID",how="outer")
```

```python
pd.merge(employees,departments,on="Dept_ID",how="left")
```

```python
pd.merge(employees,departments,on="Dept_ID",how="right")
```

```python
employee = pd.DataFrame({
    "Name": ["Amit", "Sneha", "Rahul", "Priya", "Akash"],
    "Salary": [25000, 30000, 28000, 32000, 35000]
}, index=[101, 102, 103, 104, 105])

print(employee)
```

```python
department = pd.DataFrame({
    "Department": ["IT", "HR", "Finance", "IT", "Marketing"],
    "Location": ["Pune", "Mumbai", "Nashik", "Pune", "Delhi"]
}, index=[101, 102, 103, 104, 106])

print(department)
```

```python
employee.join(department)
```

```python
employee.join(department,how="outer")
```

```python
employee.join(department,how="left")
```

```python
employee.join(department,how="right")
```

```python
department.join(employee,how="right")
```

```python
department.join(employee,how="right")
```

```python
department.join(employee,how="inner")
```

```python


data = {
    "Name": ["Amit", "Sneha", "Rahul", "Priya", "Akash",
             "Neha", "Rohit", "Pooja", "Kiran", "Anjali"],

    "Department": ["IT", "HR", "IT", "Finance", "HR",
                   "IT", "Finance", "HR", "IT", "Finance"],

    "City": ["Pune", "Mumbai", "Pune", "Nashik", "Mumbai",
             "Nashik", "Pune", "Mumbai", "Pune", "Nashik"],

    "Salary": [30000, 28000, 35000, 32000, 30000,
               40000, 38000, 27000, 45000, 36000],

    "Experience": [1, 2, 3, 2, 1, 4, 5, 1, 6, 3]
}

group_by = pd.DataFrame(data)
group_by
```

```python
#group by can not print direct data thaye having create object
da = group_by.groupby("Department")
da
```

```python
#but we can access that   object usin for loop
for i,v in da:
   print(i)
   print(v)
```

```python
group_by.groupby("Department")["Salary"].sum()
```

```python
group_by.groupby(["Department","Experience"])["Salary"].mean()
```
